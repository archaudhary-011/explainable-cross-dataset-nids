```python
#Cell 1 — load and inspect:
import sys, os
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

import pandas as pd
from src.utils import load_config

cfg = load_config()

ddos2019_path = os.path.join(cfg["project_root"], "datasets", "CICDDoS2019", "UDPLag.csv")

df_ddos2019 = pd.read_csv(ddos2019_path, low_memory=False)
df_ddos2019.columns = [str(c).strip() for c in df_ddos2019.columns]

print("Shape:", df_ddos2019.shape)
print("\nColumn names:")
for i, c in enumerate(df_ddos2019.columns, 1):
    print(f"{i:>3}. {c}")
```

```python
#Cell 2 — check the label column:
label_candidates = [c for c in df_ddos2019.columns if "label" in c.lower()]
print("Label-like columns found:", label_candidates)

label_col = label_candidates[0]
print(f"\nUsing label column: '{label_col}'")
print("\nClass distribution:")
print(df_ddos2019[label_col].value_counts())
```

```python
#Cell 3 — compare columns against our 20 selected features:
import json

with open(os.path.join(cfg["paths"]["reports_dir"], "selected_features.json")) as f:
    selection_info = json.load(f)
selected_features_2017 = selection_info["selected_features"]

cols_ddos2019 = set(df_ddos2019.columns)

matches = [f for f in selected_features_2017 if f in cols_ddos2019]
missing = [f for f in selected_features_2017 if f not in cols_ddos2019]

print(f"Exact name matches: {len(matches)}/20")
print("Matched:", matches)
print(f"\nMissing (need name mapping check): {len(missing)}/20")
print("Missing:", missing)

print("\nFull CICDDoS2019 column list (for manual comparison):")
print(list(df_ddos2019.columns))
```
