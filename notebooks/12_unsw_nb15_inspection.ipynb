```python
#Cell 1 — load and inspect both files:
import sys, os
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

import pandas as pd
from src.utils import load_config

cfg = load_config()

unsw_dir = os.path.join(cfg["project_root"], "datasets", "UNSW_NB15")

train_path = os.path.join(unsw_dir, "UNSW_NB15_training-set.csv")
test_path = os.path.join(unsw_dir, "UNSW_NB15_testing-set.csv")

df_unsw_train = pd.read_csv(train_path, low_memory=False)
df_unsw_test = pd.read_csv(test_path, low_memory=False)

print("Training set shape:", df_unsw_train.shape)
print("Testing set shape:", df_unsw_test.shape)

# Combine both - we don't need UNSW's own train/test split, since it's
# purely a CROSS-DATASET TEST TARGET for our CICIDS2017-trained model
df_unsw = pd.concat([df_unsw_train, df_unsw_test], ignore_index=True)
df_unsw.columns = [str(c).strip() for c in df_unsw.columns]

print("\nCombined shape:", df_unsw.shape)
print("\nColumn names:")
for i, c in enumerate(df_unsw.columns, 1):
    print(f"{i:>3}. {c}")
```

```python
#Cell 2 — check the label columns:
# UNSW-NB15 has TWO label-related columns: 'label' (binary 0/1) and
# 'attack_cat' (the specific attack type, or 'Normal')
print("Unique values in 'label' column:")
print(df_unsw["label"].value_counts())

print("\nUnique values in 'attack_cat' column:")
print(df_unsw["attack_cat"].value_counts())
```

```python
#Cell 3 — compare columns against our 20 selected features:
import json

with open(os.path.join(cfg["paths"]["reports_dir"], "selected_features.json")) as f:
    selection_info = json.load(f)
selected_features_2017 = selection_info["selected_features"]

print("Our 20 selected features (from CICIDS2017):")
for f in selected_features_2017:
    print(f"  - {f}")

print(f"\nUNSW-NB15 has {df_unsw.shape[1]} columns total.")
print("\nUNSW-NB15 column names (for manual comparison):")
print(list(df_unsw.columns))
```
